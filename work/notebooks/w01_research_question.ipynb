## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

54.2% of pages in the starter dataset are currently declining. At this volume, no editor could realistically review every declining page by hand, which is exactly the kind of problem a ranked, prioritized list solves. Lets the editor focus on the pages that matter most

1. Declining pages average ~4,919 impressions over 90 days which nearly identical to pages trending up (~4,716) showing decline is not confined to low-traffic pages. These are pages with real, ongoing visibility, not dead pages nobody sees.

2. Declining pages have a median search position of ~11.3 which is near the top of page 2 of search results. That's a strong, recoverable position, making these pages worth fixing rather than pages ranked too low to matter.

3. Declining pages average a CTR of ~0.32%, notably lower than flat (1.38%) or stable (0.52%) pages indicating decline shows up not just in visibility but in click efficiency too.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd

csv_path = os.path.join(project_root, "data", "raw", "content_refresh_anonymized.csv")

df = pd.read_csv(csv_path)

declining_count = (df['trend_direction'] == 'down').sum()
total_rows = len(df)
declining_pct = declining_count / total_rows * 100

print(f"Declining pages: {declining_count} out of {total_rows} ({declining_pct:.1f}%)")

avg_impressions_by_trend = df.groupby('trend_direction')['impressions_90d'].mean()
print("Average impressions_90d by trend_direction:")
print(avg_impressions_by_trend)

median_position_by_trend = df.groupby('trend_direction')['avg_position'].median()
print("Median avg_position by trend_direction:")
print(median_position_by_trend)

avg_ctr_by_trend = df.groupby('trend_direction')['ctr'].mean()
print("Average ctr by trend_direction:")
print(avg_ctr_by_trend)
```
